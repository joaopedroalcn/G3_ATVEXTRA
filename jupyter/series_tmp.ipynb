# Normalização dos sensores e preparação das janelas de dados
Normaliza os dados dos sensores, define parâmetros de janela e cria as funções para gerar as

```python
# Bibliotecas principais
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.layers import LSTM, Dense, BatchNormalization, InputLayer
from tensorflow.keras.initializers import HeNormal
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import matplotlib.pyplot as plt

# Estatísticas globais fornecidas
min_value = -37969.1569
max_value = 37128.0924

# Normalização Min-Max
sensor_cols = [col for col in data.columns if col.startswith('sensor_')]
data[sensor_cols] = (data[sensor_cols] - min_value) / (max_value - min_value)

WINDOW = 24  # ou até menor, dependendo da frequência dos eventos
STRIDE = 12

# ...existing code...
def create_windows_multisensor(data, sensor_cols, window=WINDOW, stride=STRIDE):
    X, y = [], []
    labels = data['label_binary'].values  # Troque 'target' por 'label_binary'
    sensor_data = data[sensor_cols].values
    for start in range(0, len(sensor_data) - window + 1, stride):
        end = start + window
        X.append(sensor_data[start:end, :])
        y.append(labels[start:end].max())
    return np.array(X), np.array(y)
# ...existing code...

# Usando todos os sensores
X, y = create_windows_multisensor(data, sensor_cols, window=WINDOW, stride=STRIDE)
print(X.shape, y.shape)  # (amostras, janela, sensores)
print(np.bincount(y.astype(int)))

def split_temporal(X, y, train_ratio=0.7, val_ratio=0.15):
    n = len(X)
    train_end = int(n * train_ratio)
    val_end = int(n * (train_ratio + val_ratio))
    X_train, y_train = X[:train_end], y[:train_end]
    X_val, y_val = X[train_end:val_end], y[train_end:val_end]
    X_test, y_test = X[val_end:], y[val_end:]
    return (X_train, y_train), (X_val, y_val), (X_test, y_test)

(X_train, y_train), (X_val, y_val), (X_test, y_test) = split_temporal(X, y)
print(X_train.shape, X_val.shape, X_test.shape)

# Modelo LSTM para múltiplos sensores
model = tf.keras.Sequential([
    InputLayer(input_shape=(WINDOW, len(sensor_cols))),
    LSTM(64, return_sequences=True, kernel_initializer=HeNormal()),
    BatchNormalization(),
    LSTM(64, kernel_initializer=HeNormal(), kernel_regularizer=l2(0.01)),
    BatchNormalization(),
    Dense(64, activation='relu', kernel_initializer=HeNormal(), kernel_regularizer=l2(0.01)),
    BatchNormalization(),
    Dense(1, activation='sigmoid')
])

model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=[tf.keras.metrics.Precision(), tf.keras.metrics.Recall()]
)

model.summary()
```

# Definição de callbacks e balanceamento manual dos dados e Treinamento do Modelo LSTM
Define callbacks para o treinamento, calcula pesos de classe e realiza balanceamento manual dos dados para evitar desbalanceamento.

```python
# Callbacks para evitar overfitting e ajustar o learning rate
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=1e-6
)

from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# Gera os pesos
weights = compute_class_weight('balanced', classes=np.array([0, 1]), y=y_train.astype(int))
class_weight = {0: weights[0], 1: weights[1]}

# Após criar X, y
from sklearn.utils import resample

# Separe positivos e negativos
X_pos = X[y == 1]
y_pos = y[y == 1]
X_neg = X[y == 0]
y_neg = y[y == 0]

# Oversample positivos para igualar o número de negativos (ou um valor razoável)
n_samples = min(len(y_neg), 1000)  # Limite para não explodir RAM
X_pos_upsampled, y_pos_upsampled = resample(
    X_pos, y_pos,
    replace=True,
    n_samples=n_samples,
    random_state=42
)

# Combine e embaralhe
X_balanced = np.concatenate([X_neg[:n_samples], X_pos_upsampled])
y_balanced = np.concatenate([y_neg[:n_samples], y_pos_upsampled])

# Embaralhe os dados
idx = np.random.permutation(len(y_balanced))
X_balanced = X_balanced[idx]
y_balanced = y_balanced[idx]

# Use os dados balanceados para split
(X_train, y_train), (X_val, y_val), (X_test, y_test) = split_temporal(X_balanced, y_balanced)

print(np.bincount(y_train.astype(int)))
print(np.bincount(y_balanced.astype(int)))
print(np.sum(y_train), np.sum(y_val), np.sum(y_test))

# ...modelo e callbacks permanecem iguais...

# Treine SEM class_weight (pois já balanceou manualmente)
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=32,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)


# Avaliação no conjunto de teste
test_loss, test_precision, test_recall = model.evaluate(X_test, y_test, verbose=1)
print(f'Test Loss: {test_loss:.4f} | Test Precision: {test_precision:.4f} | Test Recall: {test_recall:.4f}')

# Veja as chaves disponíveis no histórico
print(history.history.keys())
```

# (Extra) Busca de hiperparâmetros com engenharia de features e SMOTE
Executa uma busca de hiperparâmetros reduzida, adiciona features estatísticas e utiliza SMOTE para balanceamento, testando diferentes configurações do modelo.

```python
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE
from tensorflow.keras.layers import LSTM, Dense, BatchNormalization, Dropout, InputLayer
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# 1. Engenharia de Features (apenas média e diferença)
def add_stat_features(df, sensor_cols, window=12):
    for col in sensor_cols:
        df[f'{col}_mean'] = df[col].rolling(window).mean()
        df[f'{col}_diff'] = df[col].diff()
    df = df.dropna()
    return df

data = add_stat_features(data, sensor_cols, window=12)
sensor_cols = [col for col in data.columns if col.startswith('sensor_') or '_mean' in col or '_diff' in col]

# 2. Criar janelas
def create_windows_multisensor(data, sensor_cols, window=24, stride=12):
    X, y = [], []
    labels = data['label_binary'].values
    sensor_data = data[sensor_cols].values
    for start in range(0, len(sensor_data) - window + 1, stride):
        end = start + window
        X.append(sensor_data[start:end, :])
        y.append(labels[start:end].max())
    return np.array(X), np.array(y)

WINDOW = 24
X, y = create_windows_multisensor(data, sensor_cols, window=WINDOW, stride=12)

# 3. Split simples
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2, shuffle=False)

# 4. SMOTE só no treino
X_train_flat = X_train.reshape((X_train.shape[0], -1))
if len(np.unique(y_train)) > 1:
    sm = SMOTE(random_state=42)
    X_train_res, y_train_res = sm.fit_resample(X_train_flat, y_train)
    X_train_res = X_train_res.reshape((-1, WINDOW, len(sensor_cols)))
else:
    X_train_res, y_train_res = X_train, y_train

# 5. Modelo pequeno
model = tf.keras.Sequential([
    InputLayer(input_shape=(WINDOW, len(sensor_cols))),
    LSTM(32, return_sequences=True, kernel_regularizer=l2(0.001)),
    BatchNormalization(),
    Dropout(0.2),
    LSTM(16, kernel_regularizer=l2(0.001)),
    BatchNormalization(),
    Dropout(0.2),
    Dense(16, activation='relu', kernel_regularizer=l2(0.001)),
    Dense(1, activation='sigmoid')
])
model.compile(
    loss='binary_crossentropy',
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    metrics=[tf.keras.metrics.Precision(), tf.keras.metrics.Recall()]
)

early_stop = EarlyStopping(
    monitor='val_recall',
    patience=5,
    mode='max',
    restore_best_weights=True
)
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=2,
    min_lr=1e-6
)

history = model.fit(
    X_train_res, y_train_res,
    validation_data=(X_val, y_val),
    epochs=20,
    batch_size=16,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

# Avaliação final
from sklearn.metrics import classification_report, confusion_matrix
y_pred_prob = model.predict(X_val)
y_pred = (y_pred_prob > 0.5).astype(int).flatten()
print(classification_report(y_val, y_pred, digits=4))
print(confusion_matrix(y_val, y_pred))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix, precision_score, recall_score, accuracy_score, f1_score
from tensorflow.keras.layers import LSTM, Dense, Dropout, InputLayer, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import tensorflow as tf
import gc

print("=== CARREGAMENTO E ANÁLISE DOS DADOS ===")

# 1. CARREGAR TODOS OS DATASETS
datasets = []
for i in range(1, 8):
    try:
        df = pd.read_csv(f'/content/dataset_parte_{i}.csv')
        print(f"Dataset parte {i}: {df.shape}")
        datasets.append(df)
    except Exception as e:
        print(f"Erro ao carregar dataset parte {i}: {e}")

# 2. CONCATENAR TODOS OS DATASETS
data = pd.concat(datasets, ignore_index=True)
print(f"Dataset completo: {data.shape}")

# 3. ANÁLISE INICIAL DA COLUNA LABEL
print(f"\nValores únicos na coluna 'label':")
print(data['label'].value_counts(dropna=False))
print(f"Valores nulos em 'label': {data['label'].isnull().sum()}")

# 4. CONVERTER TIMESTAMP E DEFINIR COMO ÍNDICE
data['Timestamp'] = pd.to_datetime(data['Timestamp'])
data = data.sort_values('Timestamp').reset_index(drop=True)

# 5. IDENTIFICAR COLUNAS DE SENSORES
sensor_cols = [col for col in data.columns if col.startswith('sensor_')]
print(f"Sensores encontrados: {len(sensor_cols)}")
print(f"Colunas de sensores: {sensor_cols[:5]}...")  # Mostrar apenas os primeiros 5

# 6. ANÁLISE DA COLUNA LABEL - CRIAR ESTRATÉGIA BASEADA NOS DADOS REAIS
print(f"\n=== ANÁLISE DA COLUNA LABEL ===")

# Verificar tipos únicos de valores na coluna label
unique_labels = data['label'].dropna().unique()
print(f"Labels únicos (não-nulos): {unique_labels}")

# Criar label binário baseado na presença de informação
# Se label é NaN = normal (0), se tem valor = anomalia (1)
data['label_binary'] = data['label'].notna().astype(int)
print(f"Distribuição label_binary: {data['label_binary'].value_counts()}")

# 7. ANÁLISE DOS SENSORES
print(f"\n=== ANÁLISE DOS SENSORES ===")
print(f"Estatísticas dos sensores:")
sensor_stats = data[sensor_cols].describe()
print(sensor_stats)

print(f"\nRange real dos dados:")
print(f"Valor mínimo: {data[sensor_cols].min().min()}")
print(f"Valor máximo: {data[sensor_cols].max().max()}")

# Verificar valores ausentes nos sensores
missing_sensors = data[sensor_cols].isnull().sum()
print(f"\nValores ausentes por sensor:")
print(missing_sensors[missing_sensors > 0])

# 8. TRATAR VALORES AUSENTES SE HOUVER
if data[sensor_cols].isnull().sum().sum() > 0:
    print("Preenchendo valores ausentes com interpolação...")
    data[sensor_cols] = data[sensor_cols].interpolate(method='linear')
    data[sensor_cols] = data[sensor_cols].fillna(method='bfill')
    data[sensor_cols] = data[sensor_cols].fillna(method='ffill')

# 9. NORMALIZAÇÃO DOS DADOS DOS SENSORES COM VALORES ESPECÍFICOS
print(f"\n=== NORMALIZAÇÃO DOS DADOS ===")

# Valores de normalização fornecidos
min_value = -37969.1569
max_value = 37128.0924

print(f"Usando normalização com:")
print(f"Min value: {min_value}")
print(f"Max value: {max_value}")

# Aplicar normalização Min-Max com os valores específicos
data_scaled = data.copy()
data_scaled[sensor_cols] = (data[sensor_cols] - min_value) / (max_value - min_value)

print(f"Dados normalizados - Min: {data_scaled[sensor_cols].min().min():.4f}, Max: {data_scaled[sensor_cols].max().max():.4f}")

# Verificar se há valores fora do range [0,1] após normalização
out_of_range = ((data_scaled[sensor_cols] < 0) | (data_scaled[sensor_cols] > 1)).sum().sum()
if out_of_range > 0:
    print(f"ATENÇÃO: {out_of_range} valores fora do range [0,1] após normalização")
    # Clipar valores para garantir que estejam no range [0,1]
    data_scaled[sensor_cols] = np.clip(data_scaled[sensor_cols], 0, 1)
    print("Valores foram clipeados para o range [0,1]")

print(f"Verificação final - Min: {data_scaled[sensor_cols].min().min():.4f}, Max: {data_scaled[sensor_cols].max().max():.4f}")

# 10. FUNÇÃO PARA CRIAR JANELAS TEMPORAIS
def create_windows_with_context(data, sensor_cols, label_col, window=60, stride=30):
    """
    Cria janelas temporais considerando o contexto temporal
    """
    X, y = [], []

    sensor_data = data[sensor_cols].values
    labels = data[label_col].values

    for start in range(0, len(sensor_data) - window + 1, stride):
        end = start + window

        # Janela de dados dos sensores
        X.append(sensor_data[start:end, :])

        # Label da janela: 1 se houver pelo menos uma anomalia na janela
        window_labels = labels[start:end]
        y.append(1 if np.any(window_labels == 1) else 0)

    return np.array(X), np.array(y)

# 11. CRIAR JANELAS TEMPORAIS
print(f"\n=== CRIAÇÃO DAS JANELAS TEMPORAIS ===")
WINDOW_SIZE = 60  # 60 pontos por janela
STRIDE = 30       # Overlap de 50%

X, y = create_windows_with_context(data_scaled, sensor_cols, 'label_binary',
                                  window=WINDOW_SIZE, stride=STRIDE)

print(f"Janelas criadas: {X.shape}")
print(f"Distribuição das labels: Normal={np.sum(y==0)}, Anomalia={np.sum(y==1)}")

# Verificar se há valores NaN nas janelas
if np.isnan(X).sum() > 0:
    print(f"ATENÇÃO: {np.isnan(X).sum()} valores NaN encontrados nas janelas")
    X = np.nan_to_num(X, nan=0.0)
    print("Valores NaN substituídos por 0")

# 12. ESTRATÉGIA PARA CRIAR ANOMALIAS SINTÉTICAS SE NECESSÁRIO
if np.sum(y == 1) < len(y) * 0.05:  # Se menos de 5% são anomalias
    print(f"\n=== CRIANDO ANOMALIAS SINTÉTICAS ===")

    def inject_synthetic_anomalies(X_normal, n_anomalies=None):
        """Injeta anomalias sintéticas baseadas em padrões reais"""
        if n_anomalies is None:
            n_anomalies = len(X_normal) // 10  # 10% de anomalias

        X_anomalies = []

        for i in range(n_anomalies):
            # Escolher uma janela normal como base
            base_idx = np.random.randint(0, len(X_normal))
            anomaly = X_normal[base_idx].copy()

            # Aplicar diferentes tipos de anomalias
            anomaly_types = ['sensor_failure', 'spike', 'drift', 'noise_burst']
            chosen_type = np.random.choice(anomaly_types)

            if chosen_type == 'sensor_failure':
                # Falha de sensor: alguns sensores param de funcionar
                failed_sensors = np.random.choice(anomaly.shape[1],
                                                size=np.random.randint(1, 4),
                                                replace=False)
                failure_start = np.random.randint(0, anomaly.shape[0]//2)
                for sensor_idx in failed_sensors:
                    anomaly[failure_start:, sensor_idx] = 0

            elif chosen_type == 'spike':
                # Picos anômalos
                n_spikes = np.random.randint(3, 8)
                for _ in range(n_spikes):
                    spike_sensor = np.random.randint(0, anomaly.shape[1])
                    spike_time = np.random.randint(0, anomaly.shape[0])
                    spike_duration = np.random.randint(1, 5)
                    spike_magnitude = np.random.uniform(0.7, 1.0)

                    end_time = min(spike_time + spike_duration, anomaly.shape[0])
                    anomaly[spike_time:end_time, spike_sensor] = spike_magnitude

            elif chosen_type == 'drift':
                # Deriva gradual
                drift_sensors = np.random.choice(anomaly.shape[1],
                                               size=np.random.randint(2, 6),
                                               replace=False)
                for sensor_idx in drift_sensors:
                    drift_start = np.random.randint(0, anomaly.shape[0]//3)
                    drift_magnitude = np.random.uniform(0.3, 0.7)
                    drift_pattern = np.linspace(0, drift_magnitude,
                                              anomaly.shape[0] - drift_start)
                    anomaly[drift_start:, sensor_idx] += drift_pattern

            elif chosen_type == 'noise_burst':
                # Explosão de ruído
                noise_start = np.random.randint(0, anomaly.shape[0]//2)
                noise_duration = np.random.randint(5, 15)
                noise_sensors = np.random.choice(anomaly.shape[1],
                                               size=np.random.randint(3, 8),
                                               replace=False)

                end_time = min(noise_start + noise_duration, anomaly.shape[0])
                for sensor_idx in noise_sensors:
                    noise = np.random.normal(0, 0.2, end_time - noise_start)
                    anomaly[noise_start:end_time, sensor_idx] += noise

            # Garantir que os valores estejam no range [0, 1]
            anomaly = np.clip(anomaly, 0, 1)
            X_anomalies.append(anomaly)

        return np.array(X_anomalies)

    # Separar dados normais
    X_normal = X[y == 0]
    n_synthetic = min(len(X_normal) // 8, 1000)  # Máximo 1000 anomalias sintéticas

    # Criar anomalias sintéticas
    X_synthetic = inject_synthetic_anomalies(X_normal, n_synthetic)
    y_synthetic = np.ones(len(X_synthetic))

    # Combinar dados
    X_combined = np.concatenate([X, X_synthetic])
    y_combined = np.concatenate([y, y_synthetic])

    # Embaralhar
    indices = np.random.permutation(len(X_combined))
    X_combined = X_combined[indices]
    y_combined = y_combined[indices]

    print(f"Dados após anomalias sintéticas: Normal={np.sum(y_combined==0)}, Anomalia={np.sum(y_combined==1)}")

    X, y = X_combined, y_combined

# 13. SPLIT TEMPORAL DOS DADOS
def temporal_split(X, y, train_ratio=0.7, val_ratio=0.15):
    """Split temporal que preserva a ordem cronológica"""
    n = len(X)
    train_end = int(n * train_ratio)
    val_end = int(n * (train_ratio + val_ratio))

    X_train = X[:train_end]
    y_train = y[:train_end]
    X_val = X[train_end:val_end]
    y_val = y[train_end:val_end]
    X_test = X[val_end:]
    y_test = y[val_end:]

    return (X_train, y_train), (X_val, y_val), (X_test, y_test)

(X_train, y_train), (X_val, y_val), (X_test, y_test) = temporal_split(X, y)

print(f"\n=== SPLIT DOS DADOS ===")
print(f"Train: {X_train.shape}, Normal={np.sum(y_train==0)}, Anomalia={np.sum(y_train==1)}")
print(f"Val: {X_val.shape}, Normal={np.sum(y_val==0)}, Anomalia={np.sum(y_val==1)}")
print(f"Test: {X_test.shape}, Normal={np.sum(y_test==0)}, Anomalia={np.sum(y_test==1)}")

# 14. CRIAR MODELO LSTM PARA DETECÇÃO DE ANOMALIAS
print(f"\n=== CRIANDO MODELO ===")

def create_lstm_anomaly_detector(input_shape):
    model = tf.keras.Sequential([
        InputLayer(shape=input_shape),

        # Primeira camada LSTM
        LSTM(128, return_sequences=True, dropout=0.2, recurrent_dropout=0.2),
        BatchNormalization(),

        # Segunda camada LSTM
        LSTM(64, return_sequences=True, dropout=0.2, recurrent_dropout=0.2),
        BatchNormalization(),

        # Terceira camada LSTM
        LSTM(32, dropout=0.2, recurrent_dropout=0.2),
        BatchNormalization(),

        # Camadas densas
        Dense(64, activation='relu'),
        Dropout(0.4),
        BatchNormalization(),

        Dense(32, activation='relu'),
        Dropout(0.3),

        Dense(16, activation='relu'),
        Dropout(0.2),

        # Saída
        Dense(1, activation='sigmoid')
    ])
    return model

model = create_lstm_anomaly_detector((WINDOW_SIZE, len(sensor_cols)))

# 15. COMPILAR MODELO
model.compile(
    loss='binary_crossentropy',
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    metrics=['accuracy',
             tf.keras.metrics.Precision(name='precision'),
             tf.keras.metrics.Recall(name='recall'),
             tf.keras.metrics.AUC(name='auc')]
)

print("Arquitetura do modelo:")
model.summary()

# 16. CALCULAR CLASS WEIGHTS PARA BALANCEAMENTO
from sklearn.utils.class_weight import compute_class_weight

if len(np.unique(y_train)) > 1:
    class_weights = compute_class_weight('balanced',
                                       classes=np.unique(y_train),
                                       y=y_train)
    class_weight_dict = {i: class_weights[i] for i in range(len(class_weights))}
else:
    class_weight_dict = {0: 1.0, 1: 1.0}

print(f"Class weights: {class_weight_dict}")

# 17. CALLBACKS
early_stop = EarlyStopping(
    monitor='val_auc',
    patience=15,
    mode='max',
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.3,
    patience=8,
    min_lr=1e-8,
    verbose=1
)

# 18. TREINAMENTO
print(f"\n=== TREINAMENTO DO MODELO ===")
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=64,
    class_weight=class_weight_dict,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

# 19. AVALIAÇÃO
print(f"\n=== AVALIAÇÃO DO MODELO ===")
y_pred_prob = model.predict(X_test)
y_pred = (y_pred_prob > 0.5).astype(int).flatten()

# Métricas
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, zero_division=0)
recall = recall_score(y_test, y_pred, zero_division=0)
f1 = f1_score(y_test, y_pred, zero_division=0)

print(f"Acurácia: {accuracy:.4f}")
print(f"Precisão: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")

print("\nRelatório de Classificação:")
print(classification_report(y_test, y_pred, digits=4))

print("\nMatriz de Confusão:")
print(confusion_matrix(y_test, y_pred))

# 20. VISUALIZAÇÕES
plt.figure(figsize=(15, 10))

# Loss
plt.subplot(2, 3, 1)
plt.plot(history.history['loss'], label='Loss Treino')
plt.plot(history.history['val_loss'], label='Loss Validação')
plt.title('Evolução do Loss')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()
plt.grid(True, alpha=0.3)

# Accuracy
plt.subplot(2, 3, 2)
plt.plot(history.history['accuracy'], label='Acurácia Treino')
plt.plot(history.history['val_accuracy'], label='Acurácia Validação')
plt.title('Evolução da Acurácia')
plt.xlabel('Épocas')
plt.ylabel('Acurácia')
plt.legend()
plt.grid(True, alpha=0.3)

# AUC
plt.subplot(2, 3, 3)
plt.plot(history.history['auc'], label='AUC Treino')
plt.plot(history.history['val_auc'], label='AUC Validação')
plt.title('Evolução do AUC')
plt.xlabel('Épocas')
plt.ylabel('AUC')
plt.legend()
plt.grid(True, alpha=0.3)

# Precision/Recall
plt.subplot(2, 3, 4)
plt.plot(history.history['precision'], label='Precisão Treino')
plt.plot(history.history['val_precision'], label='Precisão Validação')
plt.plot(history.history['recall'], label='Recall Treino')
plt.plot(history.history['val_recall'], label='Recall Validação')
plt.title('Precisão e Recall')
plt.xlabel('Épocas')
plt.ylabel('Métrica')
plt.legend()
plt.grid(True, alpha=0.3)

# Matriz de Confusão
plt.subplot(2, 3, 5)
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Normal', 'Anomalia'],
            yticklabels=['Normal', 'Anomalia'])
plt.title('Matriz de Confusão')

# Distribuição das Probabilidades
plt.subplot(2, 3, 6)
if len(y_pred_prob[y_test == 1]) > 0:
    plt.hist(y_pred_prob[y_test == 0], alpha=0.7, label='Normal', bins=30, color='blue')
    plt.hist(y_pred_prob[y_test == 1], alpha=0.7, label='Anomalia', bins=30, color='red')
else:
    plt.hist(y_pred_prob, alpha=0.7, label='Todas as amostras', bins=30, color='blue')
plt.axvline(0.5, color='black', linestyle='--', label='Threshold=0.5')
plt.xlabel('Probabilidade Predita')
plt.ylabel('Frequência')
plt.title('Distribuição das Probabilidades')
plt.legend()

plt.tight_layout()
plt.show()

print(f"\n{'='*60}")
print("DETECÇÃO DE ANOMALIAS EM SENSORES INDUSTRIAIS")
print(f"{'='*60}")
print(f"Dataset: {len(datasets)} arquivos processados")
print(f"Normalização: Min={min_value}, Max={max_value}")
print(f"Janelas temporais: {X.shape[0]} janelas de {WINDOW_SIZE} pontos")
print(f"Sensores: {len(sensor_cols)} sensores monitorados")
print(f"Acurácia: {accuracy:.4f}")
print(f"Precisão: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")
print(f"{'='*60}")
```

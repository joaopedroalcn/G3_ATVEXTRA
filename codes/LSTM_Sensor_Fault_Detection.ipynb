### Objetivo
Implementar um modelo LSTM que identifique falhas em sinais de séries temporais coletadas de sensores, aceitando qualquer quantidade de sensores sem ser re-treinado.

### Estratégia Escolhida: Opção B - LSTM Compartilhada
- Dataset de janelas por sensor (WINDOW, 1)
- Embaralhamento de janelas de sensores distintos no batch
- Rede LSTM 1D única com pesos compartilhados
- Inferência realizada por sensor com dados ordenados temporalmente

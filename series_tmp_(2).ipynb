```python
# Bibliotecas principais
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.layers import LSTM, Dense, BatchNormalization, InputLayer
from tensorflow.keras.initializers import HeNormal
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import matplotlib.pyplot as plt

# Estatísticas globais fornecidas
min_value = -37969.1569
max_value = 37128.0924

# Normalização Min-Max
sensor_cols = [col for col in data.columns if col.startswith('sensor_')]
data[sensor_cols] = (data[sensor_cols] - min_value) / (max_value - min_value)

WINDOW = 24  # ou até menor, dependendo da frequência dos eventos
STRIDE = 12

# ...existing code...
def create_windows_multisensor(data, sensor_cols, window=WINDOW, stride=STRIDE):
    X, y = [], []
    labels = data['label_binary'].values  # Troque 'target' por 'label_binary'
    sensor_data = data[sensor_cols].values
    for start in range(0, len(sensor_data) - window + 1, stride):
        end = start + window
        X.append(sensor_data[start:end, :])
        y.append(labels[start:end].max())
    return np.array(X), np.array(y)
# ...existing code...

# Usando todos os sensores
X, y = create_windows_multisensor(data, sensor_cols, window=WINDOW, stride=STRIDE)
print(X.shape, y.shape)  # (amostras, janela, sensores)
print(np.bincount(y.astype(int)))

def split_temporal(X, y, train_ratio=0.7, val_ratio=0.15):
    n = len(X)
    train_end = int(n * train_ratio)
    val_end = int(n * (train_ratio + val_ratio))
    X_train, y_train = X[:train_end], y[:train_end]
    X_val, y_val = X[train_end:val_end], y[train_end:val_end]
    X_test, y_test = X[val_end:], y[val_end:]
    return (X_train, y_train), (X_val, y_val), (X_test, y_test)

(X_train, y_train), (X_val, y_val), (X_test, y_test) = split_temporal(X, y)
print(X_train.shape, X_val.shape, X_test.shape)

# Modelo LSTM para múltiplos sensores
model = tf.keras.Sequential([
    InputLayer(input_shape=(WINDOW, len(sensor_cols))),
    LSTM(64, return_sequences=True, kernel_initializer=HeNormal()),
    BatchNormalization(),
    LSTM(64, kernel_initializer=HeNormal(), kernel_regularizer=l2(0.01)),
    BatchNormalization(),
    Dense(64, activation='relu', kernel_initializer=HeNormal(), kernel_regularizer=l2(0.01)),
    BatchNormalization(),
    Dense(1, activation='sigmoid')
])

model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=[tf.keras.metrics.Precision(), tf.keras.metrics.Recall()]
)

model.summary()
```

```python
# Callbacks para evitar overfitting e ajustar o learning rate
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True
)
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=1e-6
)

from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# Gera os pesos
weights = compute_class_weight('balanced', classes=np.array([0, 1]), y=y_train.astype(int))
class_weight = {0: weights[0], 1: weights[1]}

# Após criar X, y
from sklearn.utils import resample

# Separe positivos e negativos
X_pos = X[y == 1]
y_pos = y[y == 1]
X_neg = X[y == 0]
y_neg = y[y == 0]

# Oversample positivos para igualar o número de negativos (ou um valor razoável)
n_samples = min(len(y_neg), 1000)  # Limite para não explodir RAM
X_pos_upsampled, y_pos_upsampled = resample(
    X_pos, y_pos,
    replace=True,
    n_samples=n_samples,
    random_state=42
)

# Combine e embaralhe
X_balanced = np.concatenate([X_neg[:n_samples], X_pos_upsampled])
y_balanced = np.concatenate([y_neg[:n_samples], y_pos_upsampled])

# Embaralhe os dados
idx = np.random.permutation(len(y_balanced))
X_balanced = X_balanced[idx]
y_balanced = y_balanced[idx]

# Use os dados balanceados para split
(X_train, y_train), (X_val, y_val), (X_test, y_test) = split_temporal(X_balanced, y_balanced)

print(np.bincount(y_train.astype(int)))
print(np.bincount(y_balanced.astype(int)))
print(np.sum(y_train), np.sum(y_val), np.sum(y_test))

# ...modelo e callbacks permanecem iguais...

# Treine SEM class_weight (pois já balanceou manualmente)
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=32,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)


# Avaliação no conjunto de teste
test_loss, test_precision, test_recall = model.evaluate(X_test, y_test, verbose=1)
print(f'Test Loss: {test_loss:.4f} | Test Precision: {test_precision:.4f} | Test Recall: {test_recall:.4f}')

# Veja as chaves disponíveis no histórico
print(history.history.keys())
```

```python
# --- Imports principais ---
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.layers import LSTM, Dense, BatchNormalization, InputLayer
from tensorflow.keras.initializers import HeNormal
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from sklearn.utils import resample
from sklearn.metrics import f1_score, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# --- Pré-processamento ---
min_value = -37969.1569
max_value = 37128.0924

sensor_cols = [col for col in data.columns if col.startswith('sensor_')]
data[sensor_cols] = (data[sensor_cols] - min_value) / (max_value - min_value)

WINDOW = 240
STRIDE = 12

def create_windows_multisensor(data, sensor_cols, window=WINDOW, stride=STRIDE):
    X, y = [], []
    labels = data['label_binary'].values
    sensor_data = data[sensor_cols].values
    for start in range(0, len(sensor_data) - window + 1, stride):
        end = start + window
        X.append(sensor_data[start:end, :])
        y.append(labels[start:end].max())
    return np.array(X), np.array(y)

def split_temporal(X, y, train_ratio=0.7, val_ratio=0.15):
    n = len(X)
    train_end = int(n * train_ratio)
    val_end = int(n * (train_ratio + val_ratio))
    return (X[:train_end], y[:train_end]), (X[train_end:val_end], y[train_end:val_end]), (X[val_end:], y[val_end:])

X, y = create_windows_multisensor(data, sensor_cols)
(X_train_orig, y_train_orig), (X_val, y_val), (X_test, y_test) = split_temporal(X, y)

# --- Balanceamento apenas no treino ---
X_pos = X_train_orig[y_train_orig == 1]
y_pos = y_train_orig[y_train_orig == 1]
X_neg = X_train_orig[y_train_orig == 0]
y_neg = y_train_orig[y_train_orig == 0]

n_samples = min(len(y_neg), 1000)
X_pos_upsampled, y_pos_upsampled = resample(X_pos, y_pos, replace=True, n_samples=n_samples, random_state=42)

X_balanced = np.concatenate([X_neg[:n_samples], X_pos_upsampled])
y_balanced = np.concatenate([y_neg[:n_samples], y_pos_upsampled])

idx = np.random.permutation(len(y_balanced))
X_train = X_balanced[idx]
y_train = y_balanced[idx]

# --- Modelo ---
model = tf.keras.Sequential([
    InputLayer(input_shape=(WINDOW, len(sensor_cols))),
    LSTM(256, return_sequences=False, kernel_regularizer=l2(0.01), kernel_initializer=HeNormal()),
    BatchNormalization(),
    Dense(1, activation='sigmoid')
])

model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=[tf.keras.metrics.Precision(name='precision'), tf.keras.metrics.Recall(name='recall')]
)

# --- Callbacks ---
early_stop = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6)

# --- Treinamento ---
history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=50,
    batch_size=32,
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

# --- Avaliação ---
y_pred_prob = model.predict(X_test)
y_pred = (y_pred_prob > 0.5).astype(int)

print(f"F1 Score: {f1_score(y_test, y_pred):.4f}")

cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Negativo", "Positivo"])
disp.plot(cmap='Blues')
plt.title("Matriz de Confusão - Teste")
plt.show()

# --- Curvas ---
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history.history['loss'], label='Treino')
plt.plot(history.history['val_loss'], label='Validação')
plt.title('Loss')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['precision'], label='Precision Treino')
plt.plot(history.history['val_precision'], label='Precision Validação')
plt.plot(history.history['recall'], label='Recall Treino')
plt.plot(history.history['val_recall'], label='Recall Validação')
plt.title('Precision / Recall')
plt.legend()
plt.show()

# --- Salvamento do modelo ---
model.save("lstm_multisensor.h5")
print("Modelo salvo como 'lstm_multisensor.h5'")
```
